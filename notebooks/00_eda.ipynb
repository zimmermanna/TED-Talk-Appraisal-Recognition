```python
import pandas as pd
```

```python
df = pd.read_csv("../data/processed/combined.csv")
dfNorm = pd.read_csv("../data/processed/combinedNorm.csv")
dfSimAgg = pd.read_csv("../data/aggregated/simAgg.csv")
dfSimAggNorm = pd.read_csv("../data/aggregated/simAggNorm.csv")
dfAdvAgg = pd.read_csv("../data/aggregated/advAgg.csv")
dfAdvAggNorm = pd.read_csv("../data/aggregated/advAggNorm.csv")
```

```python
def simple_summarize(df):
    au_cols = [c for c in df.columns if c.startswith("AU")]

    summary = pd.DataFrame({
        "Min": df[au_cols].min(),
        "Mean": df[au_cols].mean(),
        "Max": df[au_cols].max(),
        "Std": df[au_cols].std(),
    })

    return summary.round(3)
```

```python
def advanced_summarize(df):
    au_cols = [c for c in df.columns if c.startswith("AU")]

    grouped = df.groupby("source_id")[au_cols]

    summary = pd.concat({
        "Min": grouped.min(),
        "Mean": grouped.mean(),
        "Max": grouped.max(),
        "Std": grouped.std(),
        "Median": grouped.median(),
        "Q25": grouped.quantile(0.25),
        "Q85": grouped.quantile(0.85)
    }, axis=1)

    summary = (
        summary
        .stack(level=1, future_stack=True)
        .rename_axis(index=["source_id", "Action Unit"])
    )

    summary["IQR (Q85 - Q25)"] = (
        summary["Q85"] - summary["Q25"]
    )

    return summary.round(3)
```

```python
simple_summarize(df)
```

```python
advanced_summarize(df)
```

```python
simple_summarize(dfNorm)
```

```python
simple_summarize(dfSimAgg)
```

```python
simple_summarize(dfSimAggNorm)
```
